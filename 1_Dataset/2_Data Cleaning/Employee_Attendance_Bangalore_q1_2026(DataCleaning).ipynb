###Step 3 — Convert Date Columns

Many date and time columns are initially stored as text (object/string) when a CSV file is imported into Pandas. To perform date-based analysis, these columns must be converted into the datetime data type.

```python
date_columns = [
    "date_of_joining",
    "attendance_date",
    "login_timestamp",
    "logout_timestamp"
]

for col in date_columns:
    clean_df[col] = pd.to_datetime(clean_df[col], errors="coerce")
```

A list named date_columns is created to store the names of all the date and time columns. Each column in the list is then processed one by one using a for loop. The **pd.to_datetime()** function is used to convert each column from the string (object) data type to the datetime data type. The **errors='coerce'** parameter is used to ensure that any invalid or incorrectly formatted date values are converted to NaT (Not a Time) instead of causing an error. After the conversion, various date-based operations such as extracting the year, month, day, and calculating time differences can be performed.

```python
clean_df["attendance_date"].dt.month
```

This line extracts the month from every value in the attendance_date column.
<br/>
- clean_df : DataFrame
- clean_df["attendance_date"] : This selects the attendance_date column from the DataFrame.
- .dt is a Pandas datetime accessor. It allows to work with datetime columns and extract different parts of a date.
- .month extracts the month from each date.

###Step 9 — Productivity_Category

It tells us how can we classify employees based on their productive working hours

```python
def productivity_category(hours):
    if hours >= 8:
        return "Highly Productive"
    elif hours >= 6:
        return "Moderately Productive"
    else:
        return "Low Productive"

clean_df["Productivity_Category"] = clean_df["net_productive_hours"].apply(productivity_category)
```

A custom function named **productivity_category()** is created using **def**. The function receives one value **(hours)** at a time from the **net_productive_hours** column. It checks the **productive hours** using **if, elif, and else** conditions. If the hours are 8 or more, it returns Highly Productive; if the hours are between 6 and less than 8, it returns Moderately Productive; otherwise, it returns Low Productive. Finally, the **.apply()** method applies this function to every value in the **net_productive_hours** column, and the results are stored in a new column called **Productivity_Category**.
