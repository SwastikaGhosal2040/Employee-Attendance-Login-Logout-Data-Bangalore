##Step 05 — Univariate Analysis

### Step 5.1 — Distribution of Numerical Variables

The distribution of important numerical variables is examined using histograms along with Kernel Density Estimation (KDE). This helps understand the central tendency, spread, concentration, and overall shape of the numerical variables in the employee attendance dataset.

The variables considered include total hours worked, break duration, net productive hours, late arrival minutes, early exit minutes, overtime hours, and productivity ratio.

```python
# Step 5.1 — Distribution of important numerical variables

hist_cols = [
    "total_hours_worked",
    "break_duration_mins",
    "net_productive_hours",
    "late_arrival_mins",
    "early_exit_mins",
    "overtime_hours",
    "productivity_ratio"
]
# A list named hist_cols is created.
# The names of the important numerical variables to be visualized are stored in this list.
# These variables are selected for distribution analysis.

fig, axes = plt.subplots(4, 2, figsize=(14, 16))
#plt.subplots() is used to create multiple plots in one figure.
#A 4 × 2 grid is created, providing 8 subplot positions.
#fig represents the complete figure.
#axes contains the individual subplot areas.
#figsize=(14, 16) is used to specify the size of the complete figure.

axes = axes.flatten()
#The flatten() method is used to convert the 4 × 2 array of subplot axes into a one-dimensional array. This makes it easier to access each subplot using an index.

for i, col in enumerate(hist_cols): # A for loop is used to process each column stored in hist_cols. enumerate() is used to obtain both the index and the column name.
                                    # i stores the index number of the current column. col stores the name of the current column.
    sns.histplot(                   # seaborn.histplot() is used to create a histogram for the current numerical column.
        df[col].dropna(),           # The current column stored in col is selected from the DataFrame.
                                    # Missing values (NaN) are removed from the selected column before the histogram is created.
        bins=30,                    # The data is divided into 30 intervals (bins) for displaying the distribution.
        kde=True,                   # A Kernel Density Estimate (KDE) curve is added to the histogram to show the approximate shape of the data distribution.
        ax=axes[i]                  # The histogram is placed in the subplot corresponding to the current index i.
    )

    axes[i].set_title(f"Distribution of {col}")   # The title of the current subplot is set using the current column name.
    axes[i].set_xlabel(col)                       # The x-axis is labelled using the name of the current numerical column.
    axes[i].set_ylabel("Count")                   # The y-axis is labelled as Count, representing the number of observations falling within each bin.

# Remove the unused subplot
axes[-1].axis("off")         # axes[-1] refers to the last subplot. Since 7 variables are being plotted in an 8-position (4 × 2) layout, one subplot remains unused.
                             # axis("off") is used to hide the unused subplot.

plt.tight_layout()           # The spacing between the subplots is automatically adjusted so that titles, labels, and plots do not overlap.
plt.show()                   # The complete figure containing all the histograms is displayed.
```

### Step 5.2 — Boxplot Analysis of Numerical Variables

Boxplots are used to examine the distribution and variability of the selected numerical variables. They help identify the median, spread of the data, and potential outliers in employee attendance and working-hour related measures.

The analysis focuses on total hours worked, break duration, net productive hours, late arrival minutes, early exit minutes, and overtime hours.

```python
# Step 5.2 — Boxplots of important numerical variables

box_cols = [                       # A list named box_cols is created.
    "total_hours_worked",          # The names of the important numerical variables to be visualized are stored in this list.
    "break_duration_mins",         # These variables are selected for boxplot analysis.
    "net_productive_hours",
    "late_arrival_mins",
    "early_exit_mins",
    "overtime_hours"
]

fig, axes = plt.subplots(2, 3, figsize=(16, 7))    # plt.subplots() is used to create multiple plots in a single figure.
                                                   # A 2 × 3 grid is created, providing 6 subplot positions.
                                                   # fig represents the complete figure, axes contains the individual subplot areas.
                                                   # figsize=(16, 7) specifies the size of the figure.

for ax, col in zip(axes.flat, box_cols):     # A for loop is used to process each selected variable (ax,col), axes.flat provides the subplot areas one by one.
                                             # box_cols provides the column names one by one, zip() pairs each subplot with one column.
                                             # ax represents the current subplot, col represents the current column name.

    sns.boxplot(                # seaborn.boxplot() is used to create a boxplot for the current numerical column.
        x=df[col].dropna(),     # df[col] : The current column is selected from the DataFrame, .dropna() : Missing values are removed before creating the boxplot.
        ax=ax,                  # ax=ax : The boxplot is placed in the current subplot.
        color="#7fa8d9"         # A specific color(#7fa8d9) is assigned to the boxplot for visual presentation.
    )

    ax.set_title(f"Boxplot of {col}")  # The title of each subplot is created using the name of the current column.
    ax.set_xlabel(col)                 # The x-axis is labelled using the name of the current numerical variable.

plt.tight_layout()           # plt.tight_layout() is a Matplotlib function used to automatically adjust the spacing between multiple plots in a figure.
plt.show()                   # The complete figure containing all six boxplots is displayed.
```

### Step 5.3 — Statistical Summary of Numerical Variables

Descriptive statistics are calculated to summarize the important characteristics of the numerical variables. Mean, median, standard deviation, mode, skewness, and kurtosis are used to understand the central tendency, variability, and distributional shape of the variables.

```python
# Step 5.3 — Statistical summary of numerical variables

summary = {}  # An empty dictionary named summary is created. The calculated statistical values for each numerical variable will be stored in this dictionary.

for col in box_cols: # A for loop is used to process each column stored in box_cols, the current column name is temporarily stored in the variable col.

    values = df[col].dropna()  # current column is selected using df[col], .dropna() is used to remove missing values, remaining valid values stored in variable values.

    summary[col] = [           # An entry is created in the summary dictionary for the current column.
                               # The statistical values calculated below are stored as a list for that column.

        values.mean(),    # mean() function calculates the average value of the column, calculated mean is stored in the summary list.
        values.median(),  # median() function calculates the middle value of the column after the values are arranged in order, result is added to the summary list.
        values.std(),     # std() function calculates the standard deviation, indicates how much the values vary or spread around the mean.

        values.mode().iloc[0] if not values.mode().empty else np.nan,  # values.mode() is used to find the most frequently occurring value.
                                                                       # .empty checks whether the mode result is empty.
                                                                       # not values.mode().empty checks whether at least one mode is available.
                                                                   # .iloc[0] selects the first mode if a mode exists, np.nan is used if no mode is available.
        values.skew(),                                             # The skew() function calculates the skewness of the data (asymmetry of the distribution)
        values.kurt()                                              # kurt() function calculates the kurtosis of the data (shape and tail behaviour of the distribution)
    ]

summary_stats = pd.DataFrame(    # The summary dictionary is converted into a Pandas DataFrame, resulting DataFrame is stored in summary_stats.
    summary,
    index=[            # Names are assigned to the rows of the summary table, each row represents one statistical measure.
        "mean",
        "median",
        "std",
        "mode",
        "skew",
        "kurtosis"
    ]
)

# Style the table
styled_summary = (
    summary_stats.style    # The styling options of summary_stats are accessed using .style, the formatted result is stored in styled_summary
    .format("{:.6f}")      # .format() is used to control how numerical values are displayed,
                           # :.6f means that values are displayed with 6 digits after the decimal point.
    .set_table_styles([    # .set_table_styles() is used to apply formatting to different parts of the table.

        # Top row + left-most column + corner
        {
            "selector": "th",  # "th" refers to the header cells of the table, formatting is applied to these cells.
            "props": [
                ("background-color", "#e6e6e6"), # A light background colour is applied.
                ("color", "black"),              # The text colour is set to black.
                ("font-weight", "bold"),         # The header text is displayed in bold.
                ("border", "1px solid #cccccc"), # A border is added around the header cells.
                ("padding", "6px")               # Space is added inside the header cells.
            ]
        },

        # Data cells
        {
            "selector": "td",                    # "td" refers to the data cells of the table.
            "props": [                           # Borders and padding are applied to make the table easier to read.
                ("border", "1px solid #cccccc"), # A border is added around the data cells.
                ("padding", "6px")               # Space is added inside the data cells.
            ]
        },

        # Overall table
        {
            "selector": "table",                 # "table" refers to the complete table.
            "props": [
                ("border-collapse", "collapse"), # Adjacent borders are joined together for a cleaner appearance.
                ("border", "1px solid #cccccc")  # An outer border is added around the table.
            ]
        }
    ])
)

styled_summary                    # The formatted summary table is displayed as the final output.
```

###6.1 Numerical vs Numerical

#### 6.1.1 — Total Hours Worked vs Net Productive Hours

### 7.1 — Pair Plot of Important Numerical Variables

A pair plot is used to visualize the relationships and distributions among multiple numerical variables simultaneously. It helps identify correlations, trends, clusters, and potential outliers between important workplace and productivity measures.

```python
# 7.1 — Pair Plot of Important Numerical Variables
# This section is used to visualize the relationships and distributions among selected numerical variables.

import seaborn as sns                    # The Seaborn library is imported and the alias sns is assigned to it.
                                         # Seaborn is used to create statistical data visualizations.

import matplotlib.pyplot as plt          # The pyplot module of Matplotlib is imported and the alias plt is assigned to it.
                                         # It is used for creating and displaying graphs.

# Select important numerical variables
pair_cols = [                           # The selected column names are stored in the list pair_cols.
    "total_hours_worked",               # The total_hours_worked column is selected for the pair plot.
    "break_duration_mins",              # The break_duration_mins column is selected for the pair plot.
    "net_productive_hours",             # The net_productive_hours column is selected for the pair plot.
    "late_arrival_mins"                 # The late_arrival_mins column is selected for the pair plot.
]

# Use a sample for better readability
# A sample of the data is used to make the pair plot easier to read and process.

sample_df = df[pair_cols].dropna().sample(           # The selected columns are extracted from df, missing-value rows are removed,
                                                     # and a sample of the available rows is selected and stored in sample_df.

    min(800, len(df[pair_cols].dropna())),           # pair_cols : list of columns selected for the pair plot
                                                     # df[pair_cols] : The columns whose names are stored in pair_cols are selected from the DataFrame df.
                                                     # .dropna() : Rows containing missing values (NaN) are removed.

    random_state=42              # The random state is fixed at 42 so that the same rows are selected every time the code is executed.

)

# Create pair plot
# A pair plot is created to visualize the relationships between all selected numerical variables

sns.pairplot(                        # The pairplot() function from Seaborn is used to create the pair plot.

    sample_df,                       # The sampled DataFrame stored in sample_df is used as the data source for the pair plot.

    diag_kind="kde",                 # KDE (Kernel Density Estimate) plots are displayed on the diagonal
                                     # to show the distribution of each numerical variable.

    plot_kws={"alpha": 0.3, "s": 12}  # Additional settings are applied to the scatter plots.
                                      # alpha=0.3 makes the plotted points partially transparent.
                                      # s=12 sets the size of the scatter plot points.
)

plt.suptitle(                       # A common title is added to the complete pair plot.

    "Pair Plot of Important Numerical Variables",    # The title text is set to "Pair Plot of Important Numerical Variables".

    y=1.02   # The vertical position of the title is adjusted to 1.02,
             # placing the title slightly above the pair plot.
)

plt.show()   # The completed pair plot is displayed as the final output.
```

### 7.2 — Correlation Heatmap of Numerical Variables

A correlation heatmap is used to examine the strength and direction of relationships among multiple numerical variables simultaneously. It helps identify strongly correlated variables and provides an overall view of the numerical relationships in the dataset.

```python
# 7.2 — Correlation Heatmap of Numerical Variables
# A correlation heatmap is created to visualize the relationships between selected numerical variables.

import seaborn as sns                        # The Seaborn library is imported and the alias sns is assigned to it.
                                             # Seaborn is used to create statistical data visualizations.

import matplotlib.pyplot as plt              # The pyplot module of Matplotlib is imported and the alias plt is assigned to it.
                                             # It is used for creating and displaying graphs.

# Select important numerical variables :
# The numerical variables required for the correlation analysis are selected.

corr_cols = [                 # The names of the numerical columns selected for correlation analysis are stored in corr_cols.

    "total_hours_worked",     # The total_hours_worked column is selected.
    "break_duration_mins",    # The break_duration_mins column is selected.
    "net_productive_hours",   # The net_productive_hours column is selected.
    "late_arrival_mins",      # The late_arrival_mins column is selected.
    "early_exit_mins",        # The early_exit_mins column is selected.
    "overtime_hours",         # The overtime_hours column is selected.
    "tenure_days"             # The tenure_days column is selected.
]

# Calculate correlation matrix
# A correlation matrix is calculated for the selected numerical variables.

corr = df[corr_cols].corr(numeric_only=True) # A variable named corr is created to store the correlation matrix.
                                             # The columns whose names are stored in corr_cols are selected.
                                             # The corr() method is used to calculate the correlation between the selected columns.
                                             # The correlation shows the strength and direction of the linear relationship between two numerical variables.
                                             # numeric_only=True specifies that only numerical columns should be considered for the correlation calculation.
                                             # Text or categorical columns are not included in the correlation calculation.
                                             # The resulting correlation values are returned in the form of a correlation matrix.

# Plot heatmap
# A heatmap is prepared to visually represent the correlation values.

plt.figure(figsize=(9, 7)) # A new figure is created using Matplotlib.
                           # The figure is used as the drawing area for the heatmap.
                           # The figure width is set to 9 inches.
                           # The figure height is set to 7 inches.
                           # Therefore, the complete figure size is 9 × 7 inches.

sns.heatmap(            # A heatmap is created using the Seaborn library.
    corr,               # The correlation matrix stored in corr is provided as the data for the heatmap.
    annot=True,         # The numerical correlation values are displayed inside each heatmap cell.
    fmt=".2f",          # The displayed correlation values are formatted to two decimal places.
    cmap="Blues",       # A blue colour gradient is used to represent different correlation values.
    vmin=-1,            # The minimum value of the heatmap colour scale is set to -1.
    vmax=1              # The maximum value of the heatmap colour scale is set to +1.
)

plt.title("Correlation Heatmap — Numerical Variables")  # A title is added to the heatmap.
plt.tight_layout()  # The spacing between the heatmap, title, labels, and other plot elements is automatically adjusted preventing overlapping.
plt.show()          # The completed correlation heatmap is displayed as the final output.
```

## Step 08 — Hypothesis Testing

Hypothesis testing is used to determine whether observed relationships or differences in the dataset are statistically significant. The null hypothesis (H₀) assumes that there is no significant relationship or difference, while the alternative hypothesis (H₁) assumes that a significant relationship or difference exists. The appropriate statistical test is selected based on the types of variables being analyzed.

### 8.1 — Numerical vs Numerical: Pearson Correlation Test

A Pearson correlation test is performed to determine whether there is a statistically significant linear relationship between total hours worked and net productive hours.

- H₀: There is no significant linear correlation between total hours worked and net productive hours.
- H₁: There is a significant linear correlation between total hours worked and net productive hours.

```python
# 8.1 — Pearson Correlation Test
# A Pearson correlation test is performed to measure the strength and direction of the linear relationship between two numerical variables.

from scipy import stats                    # The stats module from SciPy is imported to perform statistical calculations, including the Pearson correlation test.

worked = df[                               # The complete employee DataFrame df is used for filtering the required rows.

    df["total_hours_worked"].notna() &     # The total_hours_worked column is checked for non-missing values.
                                           # .notna() checks whether a value is NOT missing.
                                           # True is returned when a value is present, and False is returned when a value is missing.

    df["net_productive_hours"].notna()     # The net_productive_hours column is checked for non-missing values.
                                           # .notna() checks whether a value is NOT missing.
                                           # True is returned when a value is present, and False is returned when a value is missing.

]

r, p = stats.pearsonr(           # The Pearson correlation function is used to calculate the linear correlation between the two selected variables.

    worked["total_hours_worked"],    # The total_hours_worked column is selected from the DataFrame stored in worked.
                                     # worked is the variable that stores the filtered DataFrame.
                                     # "total_hours_worked" is the name of the column that we want to select.
                                     # worked["total_hours_worked"] selects the total_hours_worked column from worked.


    worked["net_productive_hours"]   # The net_productive_hours column is selected from the DataFrame stored in worked.
                                     # worked is the variable that stores the filtered DataFrame.
                                     # ["net_productive_hours"] : The column named net_productive_hours is selected.
                                     # worked["net_productive_hours"] : The net_productive_hours column is selected from the filtered DataFrame stored in worked.

)

# The Pearson correlation coefficient is stored in r.
# The p-value of the correlation test is stored in p.

print(f"Pearson r = {r:.3f}")        # The print() function is used to display the result on the screen.
                                     # An f-string is used to insert the value of r into the output text.
                                     # The text "Pearson r =" is displayed before the calculated correlation value.
                                     # {r:.3f} - contains the calculated Pearson correlation value stored in r, and .3f only controls how that value is displayed.
                                     # r → Pearson correlation coefficient,
                                     # : → separates the value from its formatting instruction
                                     #.3f → formats the value as a floating-point number with 3 digits after the decimal point.


print(f"p-value = {p:.2e}")  # The print() function is used to display the result on the screen.
                             # An f-string is used to insert the calculated p-value into the output text.
                             # The text "p-value =" is displayed before the p-value.
                             # {p:.2e} represents the calculated p-value stored in p, in a formatted form.
                             # p → contains the calculated p-value obtained from stats.pearsonr().
                             # : → separates the value from its formatting instruction.
                             # .2e → displays the value in scientific notation with 2 digits after the decimal point.



if p < 0.05:                                                       # The calculated p-value is compared with the significance level of 0.05.
    print("Reject H₀ → significant linear correlation")            # The null hypothesis (H₀) is rejected because the p-value is less than 0.05.
                                                                   # A statistically significant linear correlation is indicated.
else:                                                              # This block is executed when the condition p < 0.05 is False.
                                                                   # Therefore, the p-value is greater than or equal to 0.05.
    print("Fail to reject H₀ → no significant linear correlation") # The null hypothesis is not rejected because the p-value is greater than or equal to 0.05.
                                                                   # A statistically significant linear correlation is not established.
```

### 8.2 — Numerical vs Categorical: Independent Samples t-test

An independent samples t-test is performed to determine whether the mean net productive hours differ significantly between employees working from home and employees working from the office.

- H₀: The mean net productive hours are the same for Work From Home and Work From Office employees.
- H₁: The mean net productive hours differ between Work From Home and Work From Office employees.

```python
# 8.2 — Independent Samples t-test
# An independent samples t-test is performed to compare the mean net productive hours
# between the Work From Home and Work From Office groups.

wfh = df.loc[                            # The .loc[] method is used to select specific rows and columns from df.
    df["work_mode"] == "Work From Home", # The work_mode column is checked, and only rows having "Work From Home" are selected.
    "net_productive_hours"               # From those selected rows, only the net_productive_hours column is extracted.
].dropna()                               # Any missing values (NaN) in the selected column are removed.
                                         # The final cleaned values are stored in the variable wfh.
# The non-missing net productive hours are selected for employees working from home and stored in wfh.

wfo = df.loc[                                # The .loc[] method is used to select specific rows and columns from df.
    df["work_mode"] == "Work From Office",   # only rows where the work mode is Work From Office are selected.
    "net_productive_hours"                   # "net_productive_hours" → only this column is extracted.
].dropna()                                   # .dropna() → missing values are removed.
                                             # wfo : the final values are stored in the variable wfo.
# The non-missing net productive hours are selected for Work From Office employees and stored in wfo.

# Check variance equality using Levene's test
levene_stat, levene_p = stats.levene(wfh, wfo)   # stats.levene(wfh, wfo) → Levene's test is applied to the two groups:
                                                     # wfh → Work From Home values
                                                     # wfo → Work From Office values
                                                 # levene_stat → stores the test statistic calculated by Levene's test.
                                                 # levene_p → stores the p-value produced by Levene's test.

# Use equal variance assumption only if Levene's test is not significant
equal_var = levene_p > 0.05       # levene_p contains the p-value obtained from Levene's test.
                                  # > 0.05 checks whether this p-value is greater than the significance level of 0.05.
                                  # The result will be either True or False.
                                  # This result is stored in equal_var.

t_stat, t_p = stats.ttest_ind(    # stats.ttest_ind() performs an independent samples t-test.
                                  # The calculated t-statistic is stored in t_stat.
                                  # The calculated p-value is stored in t_p.

    wfh,                 # wfh contains the net productive hours of Work From Home employees.
                         # This is provided as the first group for comparison.
    wfo,                 # wfo contains the net productive hours of Work From Office employees.
                         # This is provided as the second group for comparison.
    equal_var=equal_var  # equal_var is a parameter of ttest_ind().
                         # The value calculated earlier (True or False) is passed to it.
                         # True → equal variances are assumed.
                         # False → equal variances are not assumed.

)

print(f"Levene's test p-value = {levene_p:.4f}")  # levene_p contains the p-value obtained from Levene's test.
                                                  #:.4f means the value is displayed with 4 digits after the decimal point.
                                                  # f allows the calculated value to be inserted into the text.

print(f"t-statistic = {t_stat:.3f}") # t_stat contains the t-statistic calculated by the independent samples t-test.
                                     # .3f displays it with 3 digits after the decimal point.
print(f"p-value = {t_p:.2e}")        # t_p contains the p-value obtained from the t-test.
                                     # .2e displays the value in scientific notation with two decimal places.
print(f"WFH mean net productive hours = {wfh.mean():.2f}")  # wfh.mean() calculates the average net productive hours for Work From Home employees.
                                                            # .2f displays the result with 2 digits after the decimal point.
print(f"WFO mean net productive hours = {wfo.mean():.2f}")  # wfo.mean() calculates the average net productive hours for Work From Office employees.
                                                            # .2f displays the result with 2 digits after the decimal point.

if t_p < 0.05:                            # The t-test p-value is compared with the significance level of 0.05.
    print("Reject H₀ → net productive hours differ significantly by work mode")    # When the p-value is below 0.05, the null hypothesis is rejected
                                                                                   # A statistically significant difference between the work modes is indicated.
else:
    print("Fail to reject H₀ → no significant difference in net productive hours by work mode")
    # When the p-value is greater than or equal to 0.05, the null hypothesis is not rejected and a statistically significant difference is not established.
```

### 8.3 — Numerical vs Categorical: One-Way ANOVA

A one-way ANOVA test is performed to determine whether the mean net productive hours differ significantly across departments.

- H₀: The mean net productive hours are the same across all departments.
- H₁: At least one department has a different mean net productive hours.

```python
# 8.3 — One-Way ANOVA
# A one-way ANOVA test is performed to determine whether the mean net productive hours differ across departments.


groups = [                                     # A list named groups is created to store the net productive hours of each department separately.

    group["net_productive_hours"].dropna()     # group temporarily represents the data belonging to one department.
                                               # The net_productive_hours column is selected from the current department's data.
                                               # .dropna() is used to remove missing values (NaN).

    for _, group in df.groupby("department")  # The for loop is used to repeat the operation for each department.
                                              # The DataFrame is grouped according to the department column.
                                              # _ temporarily receives the department name, but it is not needed.
                                              # group receives the DataFrame containing the current department's data.

]                                              # The cleaned net productive hours of all departments are stored as separate groups in groups.


levene_stat, levene_p = stats.levene(*groups)    # Levene's test is performed to check whether the variances of the department groups are equal.
                                                 # The * operator is used to unpack groups so that each department group is passed separately.
                                                 # The Levene test statistic is stored in levene_stat.
                                                 # The Levene test p-value is stored in levene_p.


# One-way ANOVA
f_stat, anova_p = stats.f_oneway(*groups)        # The One-Way ANOVA test is performed to compare the mean net productive hours across departments.
                                                 # The * operator is used to unpack the groups so that each department group is passed separately to the test.
                                                 # The ANOVA F-statistic is stored in f_stat.
                                                 # The ANOVA p-value is stored in anova_p.

print(f"Levene's test p-value = {levene_p:.4f}")   # The Levene's test p-value is displayed with four decimal places.

print(f"ANOVA F = {f_stat:.2f}")                    # The calculated ANOVA F-statistic is displayed with two decimal places.

print(f"p-value = {anova_p:.2e}")                   # The ANOVA p-value is displayed in scientific notation with two decimal places.


if anova_p < 0.05:                                  # The ANOVA p-value is compared with the significance level of 0.05.

    print("Reject H₀ → net productive hours differ significantly by department")
                                                     # The null hypothesis is rejected when the p-value is less than 0.05.
                                                     # A statistically significant difference in mean net productive hours across departments is indicated.

else:                                                # This block is executed when the ANOVA p-value is greater than or equal to 0.05.

    print("Fail to reject H₀ → no significant difference in net productive hours by department")
                                                     # The null hypothesis is not rejected when the p-value is greater than or equal to 0.05.
                                                     # A statistically significant difference in mean net productive hours across departments is not established.


print("\nMean net productive hours by department:")   # The \n creates a new line, and a heading is displayed for the department-wise means.


print(
    df.groupby("department")["net_productive_hours"] # The data is grouped by department, and the net_productive_hours column is selected.
      .mean()                                        # The mean net productive hours are calculated for each department.
      .round(2)                                      # The calculated mean values are rounded to two decimal places.
      .sort_values(ascending=False)                  # The department means are sorted from highest to lowest.
)                                                    # The final department-wise mean values are displayed.
```

### 8.4 — Categorical vs Categorical: Chi-Square Test of Independence

```python
# 8.4 — Categorical vs Categorical: Chi-Square Test
# A Chi-Square test is used to check the relationship between two categorical variables.

# Department × Work Mode
# The relationship between department and work mode is being checked.


# Remove "On Leave" records because "Not Applicable" work mode
# is directly associated with being on leave
working_days = df[df["attendance_status"] != "On Leave"]       # Records other than "On Leave" are selected and stored in working_days.


# Create contingency table
contingency = pd.crosstab(
    working_days["department"],                                # The department column is used as the row variable.
    working_days["work_mode"]                                  # The work_mode column is used as the column variable.
)
                                                               # A table showing the counts for each department and work mode combination is created.


# Chi-square test of independence
chi2, chi_p, dof, expected = stats.chi2_contingency(contingency) # The Chi-Square test is performed on the contingency table.
                                                                 # The Chi-Square statistic is stored in chi2.
                                                                 # The p-value is stored in chi_p.
                                                                 # The degrees of freedom are stored in dof.
                                                                 # The expected frequencies are stored in expected.


# Display hypothesis test results
print(f"Chi-square = {chi2:.2f}")                               # The Chi-Square statistic is displayed with two decimal places.

print(f"Degrees of freedom = {dof}")                            # The degrees of freedom are displayed.

print(f"p-value = {chi_p:.4f}")                                 # The Chi-Square p-value is displayed with four decimal places.


if chi_p < 0.05:                                               # The p-value is compared with the significance level of 0.05.

    print("Reject H₀ → department and work mode are associated")
                                                               # The null hypothesis is rejected when the p-value is less than 0.05.
                                                               # A statistically significant association is indicated.

else:                                                          # This block runs when the p-value is greater than or equal to 0.05.

    print("Fail to reject H₀ → no significant association between department and work mode")
                                                               # The null hypothesis is not rejected.
                                                               # A statistically significant association is not established.


# Style the contingency table

styled_table = (                                               # A styled version of the contingency table is created and stored in styled_table.

    contingency.style                                           # The styling feature of the contingency table is accessed.

    .set_properties(**{                                         # General formatting is applied to the table cells.

        "background-color": "#FFFFFF",                          # The cell background is set to white.
        "color": "#222222",                                     # The cell text is set to dark grey.
        "font-weight": "normal",                                # Normal font weight is applied.
        "border": "1px solid #D9D9D9"                           # A light-grey border is applied to the cells.
    })

    .set_table_styles([                                         # Specific formatting rules are applied to different table parts.


        # Column headers
        {
            "selector": "th.col_heading",                       # The column header cells are selected.

            "props": [
                ("background-color", "#E8E8E8"),                # A light-grey background is applied.
                ("color", "#222222"),                           # Dark-grey text is applied.
                ("font-weight", "normal"),                      # Normal font weight is applied.
                ("border", "1px solid #D9D9D9"),                # A light-grey border is applied.
                ("text-align", "center")                        # The column header text is centered.
            ]
        },


        # Department names on the left
        {
            "selector": "th.row_heading",                       # The department row headers are selected.

            "props": [
                ("background-color", "#E8E8E8"),                # A light-grey background is applied.
                ("color", "#222222"),                           # Dark-grey text is applied.
                ("font-weight", "normal"),                      # Normal font weight is applied.
                ("border", "1px solid #D9D9D9"),                # A light-grey border is applied.
                ("text-align", "left")                          # The department names are aligned to the left.
            ]
        },


        # work_mode and department labels
        {
            "selector": "th.index_name",                        # The index-name header cells are selected.

            "props": [
                ("background-color", "#E8E8E8"),                # A light-grey background is applied.
                ("color", "#222222"),                           # Dark-grey text is applied.
                ("font-weight", "normal"),                      # Normal font weight is applied.
                ("border", "1px solid #D9D9D9")                 # A light-grey border is applied.
            ]
        },


        # Blank header cells
        {
            "selector": "th.blank",                             # The blank header cells are selected.

            "props": [
                ("background-color", "#E8E8E8"),                # A light-grey background is applied.
                ("border", "1px solid #D9D9D9")                 # A light-grey border is applied.
            ]
        },


        # Data cells
        {
            "selector": "td",                                   # The cells containing the table values are selected.

            "props": [
                ("background-color", "#FFFFFF"),                # A white background is applied.
                ("color", "#222222"),                           # Dark-grey text is applied.
                ("font-weight", "normal"),                      # Normal font weight is applied.
                ("border", "1px solid #D9D9D9"),                # A light-grey border is applied.
                ("text-align", "center")                        # The values are centered.
            ]
        },


        # Entire table
        {
            "selector": "table",                                # The entire table is selected.

            "props": [
                ("border-collapse", "collapse"),                # Adjacent borders are combined into single borders.
                ("font-size", "13px")                           # The table font size is set to 13 pixels.
            ]
        }
    ])
)

# The complete formatted contingency table is stored in styled_table.

styled_table                                                    # The formatted contingency table is displayed in the notebook.
```
